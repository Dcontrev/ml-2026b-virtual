# Vectores

## Introducción

Los vectores, tal como los usarás en este curso, son arreglos ordenados de números. En la notación, los vectores se denotan con letras minúsculas en negritas como $\mathbf{x}$. Todos los elementos de un vector son del mismo tipo. Un vector no contiene, por ejemplo, caracteres y números a la vez. El número de elementos del arreglo suele llamarse *dimensión*, aunque los matemáticos pueden preferir *rango*. El vector mostrado tiene una dimensión de $n$. Los elementos de un vector se pueden referenciar con un índice. En contextos matemáticos, los índices típicamente van de 1 a n. En ciencias de la computación y en estos laboratorios, la indexación típicamente va de 0 a n-1. En la notación, cuando se referencian individualmente los elementos de un vector, el índice se indica en un subíndice; por ejemplo, el elemento $0^{th}$ del vector $\mathbf{x}$ es $x_0$. Nota que, en este caso, la x no está en negritas.

## NumPy Arrays

La estructura de datos básica de NumPy es un *array* indexable de $n$ dimensiones que contiene elementos del mismo tipo (`dtype`). De inmediato notarás que hemos sobrecargado el término 'dimensión'. Arriba era el número de elementos del vector; aquí, dimensión se refiere al número de índices de un arreglo. Un arreglo unidimensional o 1-D tiene un índice. En nuestro curso representaremos los vectores como arreglos 1-D de NumPy. 

 - Arreglo 1-D, shape (n,): n elementos indexados de [0] a [n-1]

## Creación de Vectores

Las rutinas de creación de datos en NumPy generalmente tienen como primer parámetro el shape (forma) del objeto. Este puede ser un solo valor para un resultado 1-D o una tupla (n,m,...) que especifica el shape del resultado. A continuación se muestran ejemplos de creación de vectores con estas rutinas.

```python
# Rutinas de NumPy que reservan memoria y llenan arreglos con un valor
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

Algunas rutinas de creación de datos no reciben una tupla de shape:

```python
# Rutinas de NumPy que reservan memoria y llenan arreglos con un valor, pero no aceptan shape como argumento de entrada
a = np.arange(4.);
print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.rand(4);
print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

Los valores también se pueden especificar manualmente.

```python
# Rutinas de NumPy que reservan memoria y la llenan con valores especificados por el usuario
a = np.array([5,4,3,2]);
print(f"np.array([5,4,3,2]):  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); 
print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

Todos estos han creado un vector unidimensional `a` con cuatro elementos. `a.shape` devuelve las dimensiones. Aquí vemos que a.shape = `(4,)`, lo que indica un arreglo 1-D con 4 elementos.

## Operaciones con Vectores

### Indexado

Los elementos de los vectores se pueden acceder mediante indexado (*indexing*) y rebanado (*slicing*). NumPy ofrece un conjunto muy completo de capacidades de indexado y slicing. Aquí exploraremos solo lo básico necesario para el curso. Consulta [Slicing and Indexing](https://NumPy.org/doc/stable/reference/arrays.indexing.html) para más detalles.  
**Indexing** significa referirse a *un elemento* de un arreglo por su posición dentro del mismo.  
**Slicing** significa obtener un *subconjunto* de elementos de un arreglo con base en sus índices.  
NumPy comienza a indexar en cero, por lo que el 3er elemento de un vector $\mathbf{a}$ es `a[2]`.

```python
# operaciones de indexado sobre vectores 1-D
a = np.arange(10)
print(a)

# acceder a un elemento
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# acceder al último elemento; los índices negativos cuentan desde el final
print(f"a[-1] = {a[-1]}")

# los índices deben estar dentro del rango del vector o producirán un error
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

### Slicing

El slicing crea un arreglo de índices usando un conjunto de tres valores (`start:stop:step`). También es válido un subconjunto de esos valores. Su uso se explica mejor con ejemplos:

```python
# operaciones de slicing sobre vectores
a = np.arange(10)
print(f"a         = {a}")

# acceder a 5 elementos consecutivos (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# acceder a 3 elementos separados por dos posiciones
c = a[2:7:2];     print("a[2:7:2] = ", c)

# acceder a todos los elementos desde el índice 3 en adelante
c = a[3:];        print("a[3:]    = ", c)

# acceder a todos los elementos antes del índice 3
c = a[:3];        print("a[:3]    = ", c)

# acceder a todos los elementos
c = a[:];         print("a[:]     = ", c)
```

### Operaciones con vectores simples

Existen varias operaciones útiles que involucran un solo vector.


```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# negar los elementos de a
b = -a 
print(f"b = -a        : {b}")

# sumar todos los elementos de a; devuelve un escalar
b = np.sum(a) 
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
```

### Operaciones vectoriales elemento a elemento (_element-wise_)

La mayoría de las operaciones aritméticas, lógicas y de comparación de NumPy también se aplican a vectores. Estos operadores funcionan elemento por elemento. Por ejemplo 
$$ c_i = a_i + b_i $$

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
```

Por supuesto, para que esto funcione correctamente, los vectores deben tener el mismo tamaño:

```python
# intentar una operación entre vectores de tamaños distintos
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

### Operaciones vectoriales escalares

Los vectores se pueden 'escalar' mediante valores escalares. Un valor escalar es simplemente un número. El escalar multiplica todos los elementos del vector.

```python
a = np.array([1, 2, 3, 4])

# multiplicar a por un escalar
b = 5 * a 
print(f"b = 5 * a : {b}")
```

### Producto interior interior o punto

El producto punto (*dot product*) es un pilar del Álgebra Lineal y de NumPy. Es una operación que se utiliza ampliamente en este curso y debe comprenderse bien. A continuación se muestra el producto punto.

<img src="dot_notrans.gif" width=500>

El producto punto multiplica elemento a elemento los valores de dos vectores y luego suma el resultado.
El producto punto de vectores requiere que las dimensiones de ambos vectores sean iguales.

Implementemos a continuación nuestra propia versión del producto punto:

**Usando un ciclo for**, implementa una función que devuelva el producto punto de dos vectores. La función debe devolver, dadas las entradas $a$ y $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
Supón que `a` y `b` tienen el mismo shape.

```python
def my_dot(a, b): 
    """
   Calcula el producto punto de dos vectores
 
    Args:
      a (ndarray (n,)):  vector de entrada 
      b (ndarray (n,)):  vector de entrada con la misma dimensión que a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# prueba 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

Nota que se espera que el producto punto devuelva un valor escalar. 

Probemos las mismas operaciones usando `np.dot`. 

```python
# prueba 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

Arriba notarás que los resultados para 1-D coinciden con nuestra implementación.

### Velocidad de las operaciones Vectoriales

Utilizamos la biblioteca NumPy porque mejora la velocidad y la eficiencia en el uso de memoria. Demostrémoslo:

```python
np.random.seed(1)
a = np.random.rand(10000000)  # arreglos muy grandes
b = np.random.rand(10000000)

tic = time.time()  # capturar el tiempo de inicio
c = np.dot(a, b)
toc = time.time()  # capturar el tiempo de fin

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capturar el tiempo de inicio
c = my_dot(a,b)
toc = time.time()  # capturar el tiempo de fin

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  # eliminar estos arreglos grandes de la memoria
```

Así, la vectorización proporciona una gran aceleración en este ejemplo. Esto se debe a que NumPy aprovecha mejor el paralelismo de datos disponible en el hardware subyacente. Las GPU y las CPU modernas implementan pipelines SIMD (Single Instruction, Multiple Data), que permiten emitir múltiples operaciones en paralelo. Esto es fundamental en Machine Learning, donde los conjuntos de datos suelen ser muy grandes.

# Matrices

Las matrices son arreglos bidimensionales. Todos los elementos de una matriz son del mismo tipo. En la notación, las matrices se denotan con letras mayúsculas en negritas como $\mathbf{X}$. En este y otros laboratorios, `m` suele ser el número de filas y `n` el número de columnas. Los elementos de una matriz se pueden referenciar con un índice bidimensional. En contextos matemáticos, los números del índice típicamente van de 1 a n. En ciencias de la computación y en estos laboratorios, la indexación va de 0 a n-1. 

## Arreglos y Numpy

La estructura de datos básica de NumPy es un *array* indexable de n dimensiones que contiene elementos del mismo tipo (`dtype`). Estos se describieron anteriormente. Las matrices tienen un índice bidimensional (2-D) [m,n].

En el nuestro curso, las matrices 2-D se usan para almacenar los datos de entrenamiento. Los datos de entrenamiento son $m$ ejemplos por $n$ características (*features*), lo que forma un arreglo (m,n). El nuestro curso no realiza operaciones directamente sobre matrices, sino que normalmente extrae un ejemplo como vector y opera sobre este. A continuación revisarás: 
- creación de datos
- slicing e indexado

## Creación de una matriz

Las mismas funciones que crearon vectores 1-D crearán arreglos 2-D o n-D. Aquí hay algunos ejemplos.

A continuación, se proporciona la tupla de shape para obtener un resultado 2-D. Observa cómo NumPy usa corchetes para denotar cada dimensión. Observa además que NumPy, al imprimir, mostrará una fila por línea.

```python
a = np.zeros((1, 5))                                       
print(f"a shape = {a.shape}, a = {a}")                     

a = np.zeros((2, 1))                                                                   
print(f"a shape = {a.shape}, a = {a}") 

a = np.random.random_sample((1, 1))  
print(f"a shape = {a.shape}, a = {a}") 
```

También se pueden especificar los datos manualmente. Las dimensiones se especifican con corchetes adicionales que siguen el formato de la impresión anterior.

```python
# Rutinas de NumPy que reservan memoria y la llenan con valores especificados por el usuario
a = np.array([[5], [4], [3]]);   print(f" a shape = {a.shape}, np.array: a = {a}")
a = np.array([[5],   # También se pueden
              [4],   # separar los valores
              [3]]); # en filas distintas
print(f" a shape = {a.shape}, np.array: a = {a}")
```

## Operaciones con matrices

A continuación, vamos a explorar algunas operaciones usando matrices.

### Indexado

Las matrices incluyen un segundo índice. Los dos índices describen [fila, columna]. El acceso puede devolver un elemento o una fila/columna. Véase a continuación:

```python
# operaciones de indexado sobre matrices
a = np.arange(6).reshape(-1, 2)   # reshape es una forma conveniente de crear matrices
print(f"a.shape: {a.shape}, \na= {a}")

# acceder a un elemento
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

# acceder a una fila
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
```

Vale la pena llamar la atención sobre el último ejemplo. Acceder a una matriz especificando solo la fila devolverá un *vector 1-D*.

**Reshape**  
El ejemplo anterior usó [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) para dar forma al arreglo.  
`a = np.arange(6).reshape(-1, 2) `   
Esta línea de código primero creó un *vector 1-D* de seis elementos. Luego remodeló ese vector en un arreglo *2-D* usando el comando reshape. Pudo haberse escrito:  
`a = np.arange(6).reshape(3, 2) `  
Para llegar al mismo arreglo de 3 filas y 2 columnas.
El argumento -1 indica a la rutina que calcule el número de filas dado el tamaño del arreglo y el número de columnas.

### Slicing

El slicing crea un arreglo de índices usando un conjunto de tres valores (`start:stop:step`). También es válido un subconjunto de esos valores. Su uso se explica mejor con ejemplos:

```python
# operaciones de slicing 2-D
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

# acceder a 5 elementos consecutivos (start:stop:step)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

# acceder a 5 elementos consecutivos (start:stop:step) en dos filas
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# acceder a todos los elementos
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# acceder a todos los elementos de una fila (uso muy común)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# equivale a
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```
